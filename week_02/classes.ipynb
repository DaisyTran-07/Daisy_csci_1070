```python
class a_sample_class:
     # default/basic Constructors
     def _init_(self):
        self.var = 0

    # an instance method for printing
    def print_vars(self):
        print(self.var)


# create instance of the class
obj = a_sample_class()

obj.print_vars()
```

```python
class a_sample_class:
     # default/basic Constructors
     def _init_(self):
        self.var = 0

    # an instance method for printing
    def print_vars(self):
        print(f"first {self.var}, second {self.var}")

# create instance of the class
obj = a_sample_class()

obj.print_vars()
```
