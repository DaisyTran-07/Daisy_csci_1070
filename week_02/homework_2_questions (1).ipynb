1. (2 pts) Loop through the string and put all but the asterisks into a single string and return that.

```python
asterisk = 'My* compu*ter has go*ne craz***y an*d add*ed extra *symbols.*'
class cleanmystring:
    def __init__(self, text):
        self.text = text
    
    def remove_asterisks(self):
        result = ""
        for char in self.text:
            if char != "*":
                result += char
        return result

clean = cleanmystring(asterisk)
print(clean.remove_asterisks())
```

3. (3 pts) Create a function that checks if two words have the same characters. Make sure you use loops and proper data types to solve this. Dont use built in functions like sorted, etc.
<br /> example function call: same_characters(one="dormitory", two="dirty room")
<br /> example output: "'dormitory' and 'dirty room' have the same characters" <br />
<br /> example function call: same_characters(one="analytical", two="titanically")
<br /> example output: "'analyical' and 'titanically' have the same characters"
<br /> (Assume you'll always get strings and if you dont, return a message indicating the input is in the wrong formats) 

```python
class check_characters:
    def __init__(self, first_word, second_word):
        self.first = first_word
        self.second = second_word

    def check(self):
        if not isinstance(self.first, str) or not isinstance(self.second, str):
            return "The input is in the wrong format"
        
        for char in self.first:
            if char != " " and char not in self.second:
                return f"'{self.first}' and '{self.second}' do not have the same characters"
        for char in self.second:
            if char != " " and char not in self.first:
                return f"'{self.first}' and '{self.second}' do not have the same characters"

        return f"'{self.first}' and '{self.second}' have the same character(s)"

test1 = check_characters("dormitory", "dirty room")
print(test1.check())
test2 = check_characters("number", "cat")
print(test2.check())
```

4. (3 pts) Create a function that prints a triangle of asterisks as below. 
The function accepts n as an integer and makes the longest leg of the triangle that length, decrementing by one until you close the triangle. The function also accepts a direction for the triangle as a variable. If you pass in "base-up" it returns the first, but "base-down" returns the second.
For example, for a triangle where n = 5, the following would be produced:

```python
def triangles(n, direction):
    if direction == "base-up":
        for i in range(n, 0, -1):
            print("* " * i)

    elif direction == "base-down":
        for i in range(1, n + 1):
            print("* " * i)

print("Base-up triangle:")
triangles(5, "base-up")
print("Base-down triangle:")
triangles(5, "base-down")
```

```python
def letter_position(letter):
    alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    
    if not isinstance(letter, str) or len(letter) != 1:
        return "Input is in wrong format"

    letter_input = letter.upper()
    position = alphabet.index(letter_input) + 1

    a = b = c = d = 0
    for _ in range(position):
        a += 1
        if a > 8:
            a = 8
            b += 1
            if b > 8:
                b = 8
                c += 1
                if c > 8:
                    c = 8
                    d += 1

    return f"a = {a}, b = {b}, c = {c}, d = {d} for the letter {letter_input}"
print(letter_position("x"))
```

6. (3 pts) Write a function that returns elements greater than 10 where both first and last digits of the number are odd.
<br />example input: [2,37,53,6,103,207]
<br />example output: [37,53,103]

```python
def first_last_odds(nums):
    result = []

    for num in nums:
        if num > 10:
            first_digit = int(str(num)[0])
            last_digit = num % 10

            if first_digit % 2 != 0 and last_digit % 2 != 0:
                result.append(num)

    result.sort()
    return result
print(first_last_odds((102, 12, 50, 33, 75, 19, 93)))
```

9. (3 pts) Create a class called `Dice` that accepts a single argument called `number_of_dice`. 

Create an instance method called `roll`. 

`roll` should randomly pick a number for each die when called and print the results as a list. 

Create an instance of `Dice`. Then call `roll()` on your instance for each die. 

Hint: Inside your class create an empty list called `dice`. Use [randint](https://docs.python.org/3/library/random.html#random.randrange) function to randomly generate a function between 1 and 6

```python
from random import randint

class Dice:
    def __init__(self, num_dice):
        self.num_dice = num_dice
        self.dice = []
    
    def roll(self):
        self.dice = []
        for i in range(self.num_dice):
            self.dice.append(randint(1, 6))
        print(self.dice)


my_dice = Dice(3)
my_dice.roll()
```

BONUS: (3 pts) Create a `class` that implements a versioned key-value store with rollback capability
Requirements:
<br />set(key, value) stores a key-value pair and increments the version
<br />get(key) returns the current value for that key
<br />get(key, version) returns the value at a specific version number
<br />rollback(version) reverts the entire store to a previous version
<br />get_history(key) returns all changes to a specific key with their version numbers
After a rollback, new set() operations should branch from that version (invalidating future versions)

```python
class versioned_key_value_store:
	def __init__(self):
		self.versions = [{}]  
		self.current_version = 0
		self.history = {}  

	def set(self, key, value):
		self.current_version += 1
		store_new = self.versions[-1].copy()
		store_new[key] = value
		self.versions.append(store_new)
		
		if key not in self.history:
			self.history[key] = []

	def get(self, key, version=None):
		if version is None:
			version = self.current_version
		if version < len(self.versions):
			return self.versions[version].get(key)
		return None

	def rollback(self, version):
		if version < len(self.versions):
			self.current_version = version

	def get_history(self, key):
		return self.history.get(key, [])
```
