```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

jeopardy_df = pd.read_csv('JEOPARDY_DATA.csv')
jeopardy_df.head()
```

```python
jeopardy_df.columns = jeopardy_df.columns.str.strip()
print(jeopardy_df.columns)
```

```python
from decimal import Decimal
jeopardy_df['Value'] = jeopardy_df['Value'].replace('[\$,]', '', regex=True)  
jeopardy_df['Value'] = jeopardy_df['Value'].apply(lambda x: Decimal(x) if x not in ['', 'None'] else Decimal('0'))

round_dummies = pd.get_dummies(jeopardy_df['Round'], prefix='Round')
jeopardy_df = pd.concat([jeopardy_df, round_dummies], axis=1)
jeopardy_df = jeopardy_df.drop('Round', axis=1)

print(jeopardy_df.dtypes)

print(jeopardy_df.head())
```
