```python
#Starter code. Do NOT modify this cell
import numpy as np
import pandas as pd

# Read in dataset
df = pd.read_csv('cars.csv',delimiter=';')

# Convert each column to an np.array
car = df.Car.to_numpy()
mpg = df.MPG.to_numpy()
cylinders = df.Cylinders.to_numpy()
displacement = df.Displacement.to_numpy()
horsepower = df.Horsepower.to_numpy()
weight = df.Weight.to_numpy()
acceleration = df.Acceleration.to_numpy()
model = df.Model.to_numpy()
origin = df.Origin.to_numpy()

# create an array of headers
headers = np.array(['Car', 'MPG', 'Cylinders', 'Displacement', 'Horsepower', 'Weight','Acceleration', 'Model', 'Origin'])
```

9. (3 pts) On average are Japanese cars more fuel-efficient than US cars? Write a function that can handle this comparison.  

```python
mpg_groupby = df.groupby("Origin")["MPG"].mean()
mpg_sort = mpg_groupby.sort_values(ascending = False) 
if mpg_groupby["Japan"] > mpg_groupby["US"]:
    print("Japanese cars more fuel-efficient than US cars", mpg_sort)
else:
    print("US cars more fuel-efficient than Japanese cars", mpg_sort)
```

11. (2 pts) Create a new pandas dataframe that has the average mpg, weight, and horsepower per country of origin

```python
grouped = df.groupby("Origin")
avg_mpg = grouped["MPG"].mean()
avg_weight = grouped["Weight"].mean()
avg_horsepower = grouped["Horsepower"].mean()
avg_df = pd.DataFrame({
    "Average MPG": avg_mpg,
    "Average Weight": avg_weight,
    "Average Horsepower": avg_horsepower
})

print(avg_df)
```

Extra credit (2 pts): Write a function to find the deepest level of nested brackets
<br/>Sample Input: {"a":1,"b":2,"c":{"d":3,"e":4,"f":{"g":5,"h":6}}}
<br/>Sample Output: There are 3 nested layers

```python
current_dict = {"a":1,"b":2,"c":{"d":3,"e":4,"f":{"g":5,"h":6}}}

last_keys = []

while isinstance(current_dict, dict) and current_dict:
    keys = list(current_dict.keys())
    last_key = keys[-1] 
    last_keys.append(last_key)
    current_dict = current_dict[last_key]          

brackets_number = len(last_keys)
print(f"There are {brackets_number} nested layers")
```
