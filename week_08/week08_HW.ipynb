Q1:

```python
ALTER TABLE rental
ADD COLUMN status VARCHAR(10);
UPDATE rental
SET status = CASE
        WHEN return_date < rental_date + INTERVAL '3 days' THEN 'early'
        WHEN return_date > rental_date + INTERVAL '3 days' THEN 'late'
        ELSE 'on time'
    END;
```

Q2:

```python
SELECT ci.city, SUM(p.amount) AS total_payment
FROM payment p
JOIN customer c ON p.customer_id = c.customer_id
JOIN address a ON c.address_id = a.address_id
JOIN city ci ON a.city_id = ci.city_id
WHERE ci.city IN ('Kansas City', 'Saint Louis')
GROUP BY ci.city;
```

Q3:

```python
SELECT ca.name, COUNT(fc.film_id) AS film_count
FROM category ca
JOIN film_category fc ON ca.category_id = fc.category_id
GROUP BY ca.name;
```

Q4:

Why is there a table for category and a table for film category?
I think we have to seperate these two because in this database, the category table just lists all possible film categories, like “Action” or “Horror.” But a single film can belong to more than one category, and each category can have many films. So this film_category table can be use as a bridge between category and film so the database is cleaner and easier to spot the relationship. 

Q5:

```python
SELECT f.film_id, f.title, f.length
FROM rental r
JOIN inventory i ON r.inventory_id = i.inventory_id
JOIN film f ON i.film_id = f.film_id
WHERE r.return_date BETWEEN '2005-05-15' AND '2005-05-31';
```

I pick Query 3 and 5.

When running explain plan on Query 3 (count of films per category), PostgreSQL performs a sequential scan of the category and film_category tables, then joins them. After that, it aggregates by category name. I think this approach is fast because this table is small the query mainly counts rows. In Query 5 (films returned May 15–31, 2005), PostgreSQL scans the larger rental table to filter by return date, then joins inventory and film, returning the selected film information. Also, I think the main difference between the two queries is that Query 3 works on small tables with aggregation, while Query 5 works on larger tables with a filter and multiple joins, making it heavier to execute without limit clause. In conclusion, this shows how table size, filtering, and join type can affect the query plan and performance.
