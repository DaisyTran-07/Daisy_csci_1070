Question 3:

```python
SELECT f.title, COUNT(*) AS rental_count
FROM rental r
JOIN inventory i ON r.inventory_id = i.inventory_id
JOIN film f ON i.film_id = f.film_id 
GROUP BY f.title
ORDER BY rental_count DESC
LIMIT 10;
```

Question 4:

```python
SELECT c.customer_id, c.first_name, c.last_name,
 (SELECT SUM(p.amount)
  FROM payment p
  WHERE p.customer_id = c.customer_id) 
AS total_spent
FROM customer c
ORDER BY total_spent ASC;
```

Question 5:

```python
SELECT a.first_name || ' ' || a.last_name AS actor_name, COUNT(*) AS number_movie
FROM actor a
JOIN film_actor fa ON a.actor_id = fa.actor_id
JOIN film f ON fa.film_id = f.film_id
WHERE f.release_year = 2006
GROUP BY actor_name
ORDER BY number_movie DESC; 
```

For every customer data, it then runs a correlated subquery (but join also works in this case) on the payment table to calculate the total amount spent by that specific customer using the SUM() function. Because this subquery depends on the current customer, I think it will execute repeatedly for each customer in the table. After all the calculations, a final step will be sorting the results in ascending order based on total_spent by using ORDER BY.

Lastly, the number_movie order by highest to lowest, showing which actors appeared in the movies in 2006. 

Question 7:

```python
SELECT c.name AS genre, AVG(f.rental_rate) AS avg_rental_rate
FROM category c
JOIN film_category fc ON c.category_id = fc.category_id
JOIN film f ON fc.film_id = f.film_id
GROUP BY genre
ORDER BY avg_rental_rate DESC;
```

Question 8:

```python
SELECT c.name, COUNT(r.rental_id) AS total_rentals, 
       SUM(p.amount) AS total_sales
FROM category c
JOIN film_category fc ON c.category_id = fc.category_id
JOIN inventory i ON fc.film_id = i.film_id
JOIN rental r ON i.inventory_id = r.inventory_id
JOIN payment p ON r.rental_id = p.rental_id
GROUP BY c.name
ORDER BY total_rentals DESC
LIMIT 5;
```

Extra credit:

```python
SELECT c.name AS category,
       EXTRACT (MONTH FROM r.rental_date) AS month,
       COUNT(*) AS total_films_rented
FROM rental r
JOIN inventory i ON r.inventory_id = i.inventory_id
JOIN film_category fc ON i.film_id = fc.film_id
JOIN category c ON fc.category_id = c.category_id
GROUP BY c.name, month
ORDER BY c.name, month;
```
