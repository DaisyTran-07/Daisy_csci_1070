```python
# AUC show us measures of the probability that your model ranks a random positive example higher than a random negative one
# and how well the model seperates the two classes.

# Balanced: ROC-AUC
# Imbalanced: Precision/Recall-AUC
```

```python
numbers = [10, 20, 30, 40, 50]
def calculate_average(numbers):
    total = 0
    for number in numbers:
        total += number
    
    length = len(numbers)
    average = total/length
    return average

average = calculate_average(numbers)

print("The average is:", average)
```
